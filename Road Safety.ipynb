```python
df["Index"] = df["Index"].str.title().str.strip()
df["Accident_Severity"] = df["Accident_Severity"].str.title().str.strip()
df["Accident Date"] = df["Accident Date"].str.strip()
df["Light_Conditions"] = df["Light_Conditions"].str.title().str.strip()
df["District Area"] = df["District Area"].str.title().str.strip()
df["Road_Surface_Conditions"] = df["Road_Surface_Conditions"].str.title().str.strip()
df["Road_Type"] = df["Road_Type"].str.title().str.strip()
df["Urban_or_Rural_Area"] = df["Urban_or_Rural_Area"].str.title().str.strip()
df["Weather_Conditions"] = df["Weather_Conditions"].str.title().str.strip()
df["Vehicle_Type"] = df["Vehicle_Type"].str.title().str.strip()
```
